```python
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer, pipeline
# input text data
article = """Generative AI refers to a class of artificial intelligence models capable of
producing new content such as text, images, audio, and video. Large Language Models (LLMs)
such as GPT and LLaMA are trained on massive text corpora and can perform a wide range of
natural language tasks including translation, summarization, and question answering. These models are increasingly being deployed in industry applications ranging from customer support
to software development, transforming how humans interact with machines."""
# ---------- Text Summarization ------------
model_name = "facebook/bart-large-cnn"
sum_tokenizer = AutoTokenizer.from_pretrained(model_name)
sum_model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
# Format input and generate summary
inputs = sum_tokenizer(article, return_tensors="pt", max_length=1024, truncation=True)
summary_ids = sum_model.generate(
    inputs["input_ids"],
    max_new_tokens=45,
    min_length=20,
    do_sample=False,
    forced_bos_token_id=0
)
summary = sum_tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print("Summary:\n", summary)
qa_pipeline = pipeline("text-generation", model="Qwen/Qwen2.5-1.5B-Instruct")
question = "What are Large Language Models trained on?"
# Create a clear instruction prompt with your context
qa_prompt = f"Context: {article}\n\nQuestion: {question}\n\nAnswer concisely based only on the context:"
# Generate the answer
qa_outputs = qa_pipeline(qa_prompt, max_new_tokens=50, do_sample=False)
generated_text = qa_outputs[0]["generated_text"]
# Extract just the new answer string following our prompt
answer = generated_text[len(qa_prompt):].strip()
print("\nQuestion:", question)
print("Answer:", answer)
```
